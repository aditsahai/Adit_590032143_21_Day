#Functions

```python
def greet():
    print("hello")
greet()
```

```python
def greet_person(name):
    print("hello",name)
a=input("enter your name:")
greet_person(a)
```

```python
def add(first,second):
    return first+second
a=int(input("enter the first number: "))
b=int(input("enter the second number: "))

add(a,b)
```

```python
def square(number):
    return number**2
number=int(input("enter the number of your choice: "))
square(number)
```

```python
def welcome(name='guest'):
    return f"welcome,{name}"
print(welcome())
print(welcome("adit"))
```

```python
def rectangle_area(length,width):
    return  length*width
length=int(input("enter the lenght of the rectangle :"))
width=int(input("enter the width of the rectangle :"))

rectangle_area(length,width)
```

```python

def num_is_even(number):
    if number%2==0:
        print("the number is even")
number=int(input("enter the number: "))
num_is_even(number)
```

```python
def get_full_name(start,last):
    return f"{start.title()} {last.title()}"
start=input("enter your first name :")
last=input("enter your last name: ")

get_full_name(start,last)
```

```python
def repeat_text(text,times):
    return text*times
text=input("enter the text: ")
times= int(input("enter the no of times: "))

repeat_text(text,times)
```

```python
def calculate_total_price(price,tax_rate):
    return price + price*tax_rate
price =int(input("enter price: "))
tax_rate=int(input("enter tax rate in percentage: "))

calculate_total_price(price,tax_rate)
```
