#Lists,Tuples and Sets 

```python
fruits=['apple','banana','mango']
print(fruits)
```

```python
fruits=["apple","banana"]
fruits.append("mango")
print(fruits)
```

```python
numbers=[5,2,9,1]
numbers.sort()
print(numbers)
```

```python
items=["pen","book","bag"]
print(items[1])
print(items[:2])
```

```python
coordinates=(12.9,77.6)
print(coordinates[0],coordinates[1])
```

```python
unique_numbers=set([1,2,2,3,3])
print(unique_numbers)
```

```python
a={1,2,3}
b={3,4,5}
print(a|b)
print(a&b)
```

```python
shopping=['rice','tea','soap']
print('tea' in shopping)
```

```python
numbers=[1,2,3,4]
print(numbers+[5,6])
```

```python
values = [10,20,30]
values[1]=25
print(values)
```
