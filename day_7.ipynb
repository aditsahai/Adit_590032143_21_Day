#For Loops 

```python
for number in range(1,6):
    print(number)
```

```python
for letter in 'python':
    print(letter)
```

```python
total=0
for number in [4,7,2,9]:
    total=total+number
print(total)
```

```python
for number in range(2,11,2):
    print(number)
```

```python
names=["Asha","Ravi","Mira"]
for index,name in enumerate(names,start=1):
    print(index,name)
```

```python
for number in range(1,4):
    print(f'3 x {number} = {3 *number}')
```

```python
numbers=[1,2,3,4,5,6]
squares=[]
for i in numbers:
    a=i**2
    squares.append(a)
print(squares)
```

```python
for numbers in range(1,10):
    if numbers == 5:
        break
    print(numbers)
```

```python
for numbers in range (1,6):
    if numbers == 3:
        continue
    print(numbers)
```

```python
for row in range(1,4):
    print('*'* row )
```
