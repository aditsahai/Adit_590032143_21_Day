#Functions with collection

```python
def total_number(numbers):
    return sum(numbers)
print(total_number([2,3,5]))
```

```python
def average(numbers):
    return(sum(numbers)/len(numbers))
print(average([3,4,5]))
```

```python
def largest(numbers):
    return max(numbers)
print(largest([1,2,3]))
```

```python
def count_vowels(text):
    return sum(letter in "aeiou " for letter in text.lower())
print(count_vowels("education"))
```

```python
def make_uppercase(words):
    return[word.upper() for word in words]
print(make_uppercase(['red','blue']))
```

```python
def filter_even(numbers):
    return[number for number in numbers if number%2==0]
print(filter_even(range(1,9)))
```

```python
def word_lengths(words):
    return{word: len(word) for word in words}
print(word_lengths(["cat","python"]))
```

```python
def add_item(cart,item):
    cart.append(item)
    return cart
print(add_item(["book"],"pen"))
```

```python
def merge_scores(first,second):
    result=first.copy()
    result.update(second)
    return result
print(merge_scores({'maths':80},{"science":90}))
```

```python
def unique_sorted(values):
    return sorted(set(values))
print(unique_sorted([3,1,3,2,1]))
```
