#Dictionaries 

```python
student={'name':'Asha','age':20}
print(student)
```

```python
student={'name':'Asha'}
student['score']=82
print(student)
```

```python
prices={'pen':'10','book':'80'}
print(prices['book'])
```

```python
profile={'name':'Ravi'}
print(profile.get('city','Unknown'))
```

```python
scores={'maths':80,"english":76,"science":91}
print(scores.keys())
print(scores.values())
```

```python
cart={'rice':2,"tea":1}
cart["rice"]+=1
print(cart)
```

```python
employee={'name':'Mira',"department":"Finance"}
for key, value in employee.items():
    print(key,value)
```

```python
word="banana"
count={}
for letter in word:
    count[letter]= count.get(letter,0)+1
print(count)
```

```python
product=[
    {'name':'Pen',"price":10,"quantity":3},
    {'name':'Book','price':80,"quantity":2}

]
cart_total=sum(item['price']*item["quantity"] for item in product)
print(cart_total)
```

```python
setting={'dark_mode':False}
setting.update({'language':'english'})
print(setting)
```
